# Data columns

This notebook covers the feature and label classes from which the data columns list is formed. If user-dependent inputs are required, we added a string placeholder with "<>". Further information on the feature functions can be found in the API.

Generally, the data columns list consists of a series of feature and a single label object, i.e., its backbone has the structure:

```python
from pyanno4rt.learning.features import Feature, Label

[
    Feature(...),
    Feature(...),
    # ...
    Label(...)
]
```

We already implemented a catalogue of predefined features. The corresponding default feature objects are reported below, along with the default label object.
