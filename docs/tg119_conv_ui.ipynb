## Intro

Welcome to the <i>TG-119: Conventional UI-Based Planning</i> notebook! <br><br> In this notebook, we will showcase a beginner-friendly version of the package's UI-based functionality using data from the TG-119 standard case (available from our [Github repository's docs folder](https://github.com/pyanno4rt/pyanno4rt/tree/master/docs) as .mat-files)) for conventional treatment planning.

## Import of the relevant classes

```python
from pyanno4rt.gui import GraphicalUserInterface
```

Working with the graphical user interface (GUI) requires you to initialize an object of the class `pyanno4rt.gui._gui.GraphicalUserInterface`, which can be called parameter-free. Some advantages of the GUI are:

> <b>Easy access</b> <br>
> After initializing the GUI object and launching the main window, no more coding skills are necessary. You can simply create treatment plans with just a few clicks and inputs.

> <b>Plan management</b> <br>
> The GUI object helps you to handle multiple treatment plans at the same time. You can load, save, copy, edit or delete single plans, and also perform cross-comparisons. 

> <b>Extended functionality</b> <br>
> We use the PyQt5 front-end, which provides powerful widgets with features that add to the functionality of pyanno4rt (e.g. for visualization).

## GUI initialization

```python
gui = GraphicalUserInterface()
```

### Launch the default GUI

```python
gui.launch()
```

The main window of the GUI can be accessed with the `launch` method. In the background, this starts the PyQt5 application and initializes the front-end. There will be a loading screen ("splash screen"), and afterwards you should see the window displayed below.
> Note: the main window automatically opens on the current active screen, i.e., where the mouse pointer is located at the time of initialization.

![pyanno4rt GUI](_static/notebooks/gui_tg119_conv/empty_gui.png)

Without going into too much detail, the most important parts shall be described here:
> <b>Menu bar (upper row)</b> <br>
> Load, save, drop, select or create a (new) treatment plan, open the settings/info window, exit the application.

> <b>Composer (left column)</b> <br>
> Set the configuration/optimization/evaluation parameters.

> <b>Workflow (middle column)</b> <br>
> Configure, optimize, evaluate and visualize a treatment plan, cross-compare two treatment plans, access treatment plan data (parameters, internal data, log files), export the treatment plan to a Python file.

> <b>Viewer (right column)</b> <br>
> Preview the CT/dose slices and the dose-volume histogram (DVH) on the dose grid.

> <b>Status bar (bottom row)</b> <br>
> Get the current status of the treatment plan, stop the running process, open the pyanno4rt websites (PyPI, Read the Docs, Github).

### Launch the GUI with initial treatment plan(s)

The `launch` method has an optional parameter which allows to pass one or more initial treatment plan(s). To demonstrate this, we quickly set up the treatment plan from the code-based example notebook for the TG-119 case.

```python
from pyanno4rt.base import (
    Configuration, Evaluation, Optimization, TreatmentPlan)
from pyanno4rt.optimization.components import (
    SquaredDeviation, SquaredOverdosing)
```

```python
tp = TreatmentPlan(
    configuration=Configuration(
        label='TG-119', modality='photon', imaging_path='./TG_119_data.mat',
        dose_matrix_path='./TG_119_photonDij.mat', dose_resolution=[6, 6, 6]),
    optimization=Optimization(
    components=[
        SquaredOverdosing(segment='Core', maximum_dose=25, weight=100),
        SquaredDeviation(segment='OuterTarget', target_dose=60, weight=1000),
        SquaredOverdosing(segment='BODY', maximum_dose=30, weight=800)]),
    evaluation=Evaluation())
```

Following that, we can provide the `launch` method with the treatment plan (or, alternatively, a list of plans):

```python
gui.launch(tp)
```

There are three options for setting up a treatment plan in the main window of the GUI:

In the example video, we assume that we need to create a new treatment plan using the third option. For this, we apply the following steps:

1) Open the plan creator by clicking the selector item
2) Provide the required input data (label, modality, imaging path, dose matrix path, dose grid resolution and optimization components)
3) Click the "Create" button to transfer the treatment plan to the main window
4) Optionally: adapt the treatment plan by updating the parameter groups
5) Continue with the workflow steps

> Note: in the example video, we import the imaging data and dose-influence matrix file from a local folder. Since the GUI transforms the paths into absolute values, these will look different for you.

## Treatment plan workflow

The UI-based interface exposes the configuration, optimization, evaluation and visualization methods of each treatment plan in the middle column. In the following, we guide you step-by-step through the individual workflow steps.

### Configuring the plan

<video width="100%"
       src='_static/notebooks/gui_tg119_conv/configuration' 
       controls>
</video>

Once initialized, the status bar at the bottom indicates that the treatment plan is ready to be configured. You can trigger the configuration process by clicking on the "CONFIGURE" button. Afterwards, the viewer shows the CT slices with the segment contours which you can analyze by changing the slice/plane, and the tool buttons at the bottom of the workflow column reveal the treatment plan data and logs.

### Modeling for the plan

If any machine learning outcome models were present, the "MODEL" button would become available. There are no such models in the conventional TG-119 planning case, so this button remains disabled and we can go directly to optimization (as indicated by the status bar). Check out the "TG-119: UI-Based Planning with Machine Learning Outcome Models" notebook to learn more about machine learning outcome model-based optimization in pyanno4rt.

### Optimizing the plan

<video width="100%"
       src='_static/notebooks/gui_tg119_conv/optimization' 
       controls>
</video>

By clicking the "OPTIMIZE" button in the workflow column, you start the optimization process, including preprocessing steps and the execution of the solver. The viewer will automatically update the CT/segmentation slices with an overlay of the dose distribution, whose colormap can be varied by right-clicking on the colorbar.

### Evaluating the plan

<video width="100%"
       src='_static/notebooks/gui_tg119_conv/evaluation' 
       controls>
</video>

To evaluate the optimized treatment plan, you need to click the "EVALUATE" button in the workflow column. This will calculate the dosimetric plan quality indicators and the DVH, which is then displayed on the right-hand side as an interactive widget where you can e.g. select individual curves to get basic segment information.

### Visualizing the plan

![pyanno4rt GUI](_static/notebooks/gui_tg119_conv/component_graph.png)

Finally, the "VISUALIZE" button opens the visualization window, which provides interactive plots for analysis and static plots for publication. Here, the availability of the individual plots depends on the current state of the treatment plan (initialized, configured, modeled/optimized, or evaluated). You may extend the visualization tools according to your needs (and we would love to hear about your ideas)!

### Comparing treatment plans

<video width="100%"
       src='_static/notebooks/gui_tg119_conv/comparison' 
       controls>
</video>

The main window of the GUI also features a small tool for treatment plan comparison. For demonstration purposes, we create a duplicate ("TG-119-2") of the TG-119 plan where we increase the target dose from 63 to 80 Gy.

> Note: any changes to the treatment plan parameters in the composer widget need to be saved by clicking the "Update" button at the bottom, otherwise they will not be effective.

Treatment plans for comparison must be configured, optimized and evaluated, therefore we first cycle through the workflow steps. After the status bar signals that "TG-119-2" is ready, we switch to the plan comparison tab in the workflow column. The baseline treatment plan will always be the current active plan, while the reference plan can be selected from those ready for comparison. By clicking on the "COMPARE" button, the plan comparison tool pops up:

![pyanno4rt GUI](_static/notebooks/gui_tg119_conv/plan_comparison_tool.png)

The plan comparison tool includes synchronized CT/dose slice images and the corresponding DVHs from the baseline, reference and difference dose cubes (on the CT grid), along with control elements known from the viewer widget. Additionally, you can turn on/off the CT, dose, segment and dose contours, as well as open a joint DVH and (if available) NTCP plot.

### Fetching treatment plans from the GUI

```python
tps_gui = gui.fetch()
```

The GUI has an internal dictionary in which objects of the <i>TreatmentPlan</i> class generated from the interface are stored. These can also be retrieved after closing the GUI using the <i>fetch</i> method.
