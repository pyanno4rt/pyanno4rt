# Data columns

This notebook covers the feature and label classes from which the data columns list is formed. If user-dependent inputs are required, we added a string placeholder with "<>". Further information on the feature functions can be found in the API.

Generally, the data columns list consists of a series of feature and a single label object, i.e., its backbone has the structure:

```python
from pyanno4rt.learning.features import DynamicFeature, Label, StaticFeature

[
    DynamicFeature(...),
    StaticFeature(...),
    # ...
    Label(...)
]
```

We distinguish two types of input features:
<ul>
    <li><b>Dynamic:</b> "endogenous" features associated with a segment and a (re-)calculation function, e.g. dosiomics and radiomics</li>
    <li><b>Static:</b> exogenous "contextual" features associated with a user-definable, constant value, e.g. demographics</li>
</ul>
pyanno4rt is already equipped with a compact catalogue of predefined features. The corresponding default feature objects are reported below, along with the default label object.

<i>A feature for the standard deviation of the dose</i>
