## Machine learning model-based components

### Decision Tree Outcome
<i>A class for the decision tree-based outcome</i>

```python
from pyanno4rt.learning import ModelParameters
from pyanno4rt.optimization.components import (
    DecisionTreeNTCP, DecisionTreeTCP) # depending on outcome type
from pyanno4rt.learning.features import (
    DynamicFeature, Label, StaticFeature) # depending on data column types
from pyanno4rt.learning.tune_spaces import TuneSpaceDT

DecisionTreeOutcome(
    segment='<your_segment>',
    outcome_type='NTCP',  # 'TCP'
    model=DecisionTree(
        label='<your_model_label>',
        dataset=TabularDataset(  # ImageDataset
            path='<your_data_path>',
            columns='<your_data_columns>'),
        preprocessor=TabularPreprocessor(  # ImagePreprocessor
            steps='<your_preprocessing_steps>'),
        tuner=BayesHPTuner(  # GridHPTuner, RandomHPTuner
            space=TuneSpaceDT(
                criterion=['gini', 'entropy'],
                splitter=['best', 'random'],
                max_depth=[5],
                min_samples_split=[0.0, 1.0],
                min_samples_leaf=[0.0, 0.5],
                min_weight_fraction_leaf=[0.0, 0.5],
                max_features=[0],
                class_weight=[None, 'balanced'],
                ccp_alpha=[0.0, 1.0]),
            evaluations=25,
            score='AUC',
            splits=5,
            repeats=1),
        inspector=ModelInspector(),
        evaluator=ModelEvaluator(
            splits=5,
            repeats=1),
        path='your_model_path>'),
    component_type='objective',
    embedding='active',
    weight=1,
    rank=1,
    bounds=None,
    identifier=None)
```
